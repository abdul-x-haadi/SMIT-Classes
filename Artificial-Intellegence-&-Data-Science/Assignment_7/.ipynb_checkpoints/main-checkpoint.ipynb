```python
day_q6 = int(input("Input a number to see what day is today : "))
if day_q6>0 and day_q6<=7:
    if day_q6==1:
        print("Sunday")
    elif day_q6==2:
        print("Monday")
    elif day_q6==3:
        print("Tuesday")
    elif day_q6==4:
        print("Wednesday")
    elif day_q6==5:
        print("Thursday")
    elif day_q6==6:
        print("Friday")
    elif day_q6==7:
        print("Saturday")
else:
    print("Enter number between 1-7")
```

7. Write a program that prompts the user to enter their weight (in kilograms) and height
(in meters). The program should calculate the Body Mass Index (BMI) using the formula: BMI = weight / (height * height). The program should then classify the BMI into one of
the following categories:
- less than 18.5 - Underweight
- BMI between 18.5 and 24.9 - Normal weight
- BMI between 25 and 29.9 - Overweight
- BMI 30 or greater - Obesity 

```python
weight_q7 = float(input("Enter your weight (in kgs): "))
height_q7 = float(input("Enter your height (in meters): "))
bmi_q7 = weight_q7/(height_q7**2)
if bmi_q7<18.5:
    print("Underweight")
elif bmi_q7<=24.9:
    print("Normal Weight")
elif bmi_q7<=29.9:
    print("Overweight")
else:
    print("Obesity")
```

gets a grade as per the following rules:

- 70-79 : C
- 60-69 : D

9. The roots of the quadratic equation ax2 + bx + c = 0, a ≠ 0 are given by the quadratic formula.  
In this formula, the term
b2 - 4ac is called the discriminant. If b2 - 4ac = 0, then the
equation has two equal roots.
If b2 - 4ac > 0, the equation has two real roots. If b2 - 4ac < 0, the equation has two complex roots.
Write a program that prompts the user to input the value of a (the coefficient of x2),
b(the coefficient of x), and c (the constant term) and outputs the roots of the quadratic
equation. 


```python
a = int(input("Enter value of a (the coefficient of x2) : "))
b = int(input("Enter value of b (the coefficient of x)"))
c = int(input("Enter the value of c (the constant term) : "))
if ((b**2)-(4*a*c))==0:
    print("Two equal roots")
elif ((b**2)-(4*a*c))>0:
    print("Two real roots")
else :
    print("Two complex roots")
```

12. Write a program that prompts the user to input a character and determine the
character is vowel or consonant. 


```python
v_or_c = input("Enter a alphabet").lower().strip()
if len(v_or_c)==1:
    if v_or_c=="a" or v_or_c=="e" or v_or_c=="i" or v_or_c=="o" or v_or_c=="u":
        print(v_or_c, " is a vowel")
    else:
        print(v_or_c, " is a consonant.")
else: 
    print("Enter one character ONLY")
```

13. Write a program that prompts the user to input a year and determine whether the
year is a leap year or not.
Leap Years are any year that can be evenly divided by 4. A year that is evenly divisible by
100 is a leap year only if it is also evenly divisible by 400. Example:
- 1992 Leap Year
- 2000 Leap Year
- 1900 NOT a Leap Year
- 1995 NOT a Leap Year

```python
year_q13 = int(input("Enter a year : "))
if year_q13%4==0 or (year_q13%100==0 and year_q13%400==0):
    print("Leap Year")
else :
    print("Not a leap year")
```

14. Write a program that prompts the user to input number of calls and calculate the
monthly telephone bills as per the following rule:
Minimum Rs. 200 for up to 100 calls.
- Plus Rs. 0.60 per call for next 50 calls.
- Plus Rs. 0.50 per call for next 50 calls.
- Plus Rs. 0.40 per call for any call beyond 200 calls. 

```python
num_calls = int(input("Enter number of calls you did this months : "))
if num_calls<=50:
    print(f"Telephone Bill : Rs. {num_calls*0.6}")
elif num_calls<=100:
    secondcalls = (num_calls-50)*0.5
    print(f"Telephone Bill : Rs. {(50*0.6)+secondcalls}")
elif num_calls>=200:
    print(f"Telephone Bill : Rs. {num_calls*0.4}")
```
