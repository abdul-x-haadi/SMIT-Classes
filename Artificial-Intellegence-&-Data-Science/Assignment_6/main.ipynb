### A- Discounts and Pricing

Write a program to check if a customer is eligible for a discount. If the total purchase is
more than $100, apply a 10% discount and display the final price. Otherwise, display the
total price as it is.

```python
eligibility = int(input("Enter amount you spent : $"))
if eligibility>=100 :
    print(f"You got 10% percent discount. Your final amount is : ${eligibility-(eligibility*0.1)}")
else : 
    print(f"Your final amount is {eligibility}")
```

### B- Tax Calculations

If the price of an item is greater than $500, apply a luxury tax of 15%. Otherwise, apply a
standard tax of 8%. Display the total price after tax.

```python
price_q6 = int(input("Enter amount you spent : $"))
if price_q6>=500:
    print(f"After tax(luxury) applied your total payment is : ${price_q6+(price_q6*0.15)}")
else : 
    print(f"After tax(standard) applied your total payment is : ${price_q6+(price_q6*0.08)}")
```
