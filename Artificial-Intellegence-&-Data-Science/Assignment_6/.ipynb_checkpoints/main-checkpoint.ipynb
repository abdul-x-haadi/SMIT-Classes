### A- Discounts and Pricing

Write a program to check if a customer is eligible for a discount. If the total purchase is
more than $100, apply a 10% discount and display the final price. Otherwise, display the
total price as it is.

```python
eligibility = int(input("Enter amount you spent : $"))
if eligibility>=100 :
    print(f"You got 10% percent discount. Your final amount is : ${eligibility-(eligibility*0.1)}")
else : 
    print(f"Your final amount is {eligibility}")
print("--------------------------------------------------------------------------------------------------")
```

If a customer buys more than 5 items, apply a 15% discount on the total price. Otherwise,
no discount is applied. Display the total price.

```python
total_itmes = int(input("How many items did you buy ? \n"))
total_price = int(input("Enter amount you spent : $"))
if total_itmes>=5 :
    print(f"You got 15% descount. Your total price now is : ${total_price-(total_price*0.15)}")
else : 
    print(f"No discount is applied. Your total price is : ${total_price}")
print("--------------------------------------------------------------------------------------------------")
```

Check if the customer is a member (is_member = True). Members get a 20% discount;
non-members get a 5% discount. Calculate and print the discounted price.

```python
is_member = input("Are you a member  : (y/n)").lower()
price = int(input("Enter amount you spent : $"))
if is_member == "y":
    print(f"You got 20% discount. Your total amount is ${price-(price*0.2)}")
else : 
    print(f"You got 5% discount. Your total amount is ${price-(price*0.05)}")
print("--------------------------------------------------------------------------------------------------")
```

If today is a holiday (is_holiday = True), apply a 25% discount; otherwise, apply a 10%
discount. Calculate the price after discount.

```python
is_holiday = input("Is today Holiday? : (Y/N)").lower().strip()
price_q4 = int(input("Enter amount you spent : $"))
if is_holiday == "y" :
    print(f"You got 25% discount. Your total amount is ${price_q4-(price_q4*0.25)}")
else:
    print(f"You got 10% discount. Your total amount is ${price_q4-(price_q4*0.1)}")
print("--------------------------------------------------------------------------------------------------")
```

If a customer buys an even number of items, they get half of them for free. Otherwise,
they pay for all. Calculate the number of items the customer has to pay for.

```python
items_q5 = int(input("Enter number of items you bought : "))
price_q5 = int(input("Enter amount you spent : $"))
if items_q5%2==0 :
    print(f"You have to pay half of the price. Your total price is ${price_q5/2}")
else : 
    print(f"Your total price is ${price_q5}")
print("--------------------------------------------------------------------------------------------------")
```

### B- Tax Calculations

If the price of an item is greater than $500, apply a luxury tax of 15%. Otherwise, apply a
standard tax of 8%. Display the total price after tax.

```python
price_q6 = int(input("Enter amount you spent : $"))
if price_q6>=500:
    print(f"After tax(luxury) applied your total payment is : ${price_q6+(price_q6*0.15)}")
else : 
    print(f"After tax(standard) applied your total payment is : ${price_q6+(price_q6*0.08)}")
print("--------------------------------------------------------------------------------------------------")
```

If a person's annual income is above $50,000, they pay 20% tax. Otherwise, they pay
10%. Calculate and display the tax amount.

```python
income_q7 = int(input("Enter your income : $"))
if income_q7 >= 50000:
    print(f"Pay 20% tax : ${income_q7*0.2}")
else : 
    print(f"Pay 10% tax : ${income_q7*0.1}")
print("--------------------------------------------------------------------------------------------------")
```

Write a program to categorize a person into tax brackets:
- Income < 30,000: "Low Tax"
- 30,000 ≤ Income < 100,000: "Medium Tax"
- Income ≥ 100,000: "High Tax"

```python
income_q8 = int(input("Enter your income : $"))
if income_q8<30000:
    print("Low Tax")
elif income_q8 >= 30000 and income_q8<100000:
    print("Medium Tax")
else : 
    print("High Tax")
print("--------------------------------------------------------------------------------------------------")
```

### D- Travel and Tickets

If a person is traveling more than 500 miles, offer a 20% discount on ticket price. Otherwise, charge the full amount.

```python
travel_dist = int(input("Enter travelling distance : "))
ticket_price = int(input("Enter ticket price : $"))
if travel_dist>=500:
    print(f"You got 20% discount. Ticket Price : ${ticket_price - (ticket_price*0.2)}")
else : 
    print(f"Ticket Price : ${ticket_price}")
print("--------------------------------------------------------------------------------------------------")
```

If a passenger is under 12 or over 60 years old, apply a 15% discount on the ticket price.
Otherwise, charge the full price.

```python
ticket_price_q17 = int(input("Enter ticket price : $"))
age_q17 = int(input("Enter your age : "))
if age_q17<=12 or age_q17>=60 : 
    print(f"15% discount. Final Ticket Price : ${ticket_price_q17 - (ticket_price_q17*0.15)}")
else : 
    print(f"Final Ticket Price : ${ticket_price_q17}")
print("--------------------------------------------------------------------------------------------------")
```

If the ticket is for a weekend (is_weekend = True), add a 10% surcharge. Otherwise,
charge the standard price

```python
is_weekend_q18 = input("Is taday weekend : (y/n) : ").lower()
ticket_price_q18 = int(input("Enter ticket price : $"))
if is_weekend_q18 == "y" : 
    print(f"10% surcharge. Final Price : ${ticket_price_q18 + (ticket_price_q18*0.1)}")
else : 
    print(f"Final Price : ${ticket_price_q18}")
print("--------------------------------------------------------------------------------------------------")
```

### E- Grades and Performance

If a student scores 40 or more, print "Pass". Otherwise, print "Fail"

```python
score = int(input("Enter your score : "))
if score>=40:
    print("Pass")
else : 
    print("Fail")
print("--------------------------------------------------------------------------------------------------")
```

Based on a student's score, assign grades:
- 90 and above: "A"
- 75–89: "B"
- 50–74: "C"
- Below 50: "F"

```python
score_q22 = int(input("Enter your score  : "))
if score_q22>=0 and score_q22<=100:
    if score_q22>=90:
        print("A")
    elif score_q22>=75:
        print("B")
    elif score_q22>=50:
        print("C")
    else:
        print("F")
else:
    print("Enter appropiate marks.")
print("--------------------------------------------------------------------------------------------------")
```
