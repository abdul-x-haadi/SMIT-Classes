#### 1) Area of a Rectangle: 

```python
length1 = int(input("Enter length of rectangle : "))
width1 = int(input("Enter width of rectangle :"))
area = length1*width1
print(f"""The area of rectangle is :{area}""")
```

#### 3) Simple Interest: 

```python
principal3 = int(input("Enter principal for simple interest : "))
rate3 = int(input("Enter rate for simple interest : "))
time3 = int(input("Enter Time for simple interest"))
simple_interest3 = principal3*rate3*time3
print(f"The simple interest is {simple_interest3}")
```

#### 7) Compound Interest: 

```python
p7 = int(input("Enter principal amount : "))
r7 = float(input("Enter annual interest rate (decimal) : "))
n7 = int(input("Enter number of times interest is compounded per year :"))
t7 = float(input("Enter time in years :"))
a7 = p7*(1+(r7/n7))**(n7-t7)
print(f"Compound Interest is : {a7}")
```

#### 11) Quadratic Equation Roots: 

```python
a11 = int(input("a of quadratic equation : "))
b11 = int(input("b of quadratic equation : "))
c11 = int(input("c of quadratic equation : "))
q_e11_1 = (-b11+((b11**2)-(4*a11*c11))**(1/2))/(2*a11)
q_e11_2 = (-b11-((b11**2)-(4*a11*c11))**(1/2))/(2*a11)
print(f"The answer of quadratic equation is {q_e11_1} and {q_e11_2}")
```
